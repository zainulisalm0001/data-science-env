```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_log_error

def compute_rmsle(y_test: np.ndarray, y_pred: np.ndarray, precision: int = 3) -> float:
    """RMSLE as required by assignment"""
    rmsle = np.sqrt(mean_squared_log_error(y_test, y_pred))
    return round(rmsle, precision)
```

```python

target = "SalePrice"


cont_features = ["MSSubClass", "LotFrontage"]   
cat_features  = ["MSZoning", "Street"]         
features = cont_features + cat_features
```

```python
X = df[features].copy()
y = df[target].copy()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

print("X_train:", X_train.shape, "X_test:", X_test.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder


num_imputer = SimpleImputer(strategy="median")
scaler = StandardScaler()

X_train_num = num_imputer.fit_transform(X_train[cont_features])
X_train_num_scaled = scaler.fit_transform(X_train_num)

X_test_num = num_imputer.transform(X_test[cont_features])
X_test_num_scaled = scaler.transform(X_test_num)


cat_imputer = SimpleImputer(strategy="constant", fill_value="Missing")
X_train_cat = cat_imputer.fit_transform(X_train[cat_features])
X_test_cat  = cat_imputer.transform(X_test[cat_features])

import sklearn

if int(sklearn.__version__.split(".")[1]) >= 2: 
    ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
else:
    ohe = OneHotEncoder(handle_unknown="ignore", sparse=False)

X_train_cat_ohe = ohe.fit_transform(X_train_cat)
X_test_cat_ohe  = ohe.transform(X_test_cat)


cat_feature_names = ohe.get_feature_names_out(cat_features)


X_train_cat_df = pd.DataFrame(X_train_cat_ohe, columns=cat_feature_names, index=X_train.index)
X_test_cat_df  = pd.DataFrame(X_test_cat_ohe,  columns=cat_feature_names, index=X_test.index)


X_train_prep = pd.DataFrame(X_train_num_scaled, columns=cont_features, index=X_train.index).join(X_train_cat_df)
X_test_prep  = pd.DataFrame(X_test_num_scaled, columns=cont_features, index=X_test.index).join(X_test_cat_df)

print("Prepared X_train:", X_train_prep.shape)
print("Prepared X_test :", X_test_prep.shape)
```

```python

y_train_log = np.log1p(y_train)

model = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
model.fit(X_train_prep, y_train_log)
```

```python
y_pred_log = model.predict(X_test_prep)
y_pred = np.expm1(y_pred_log)

rmsle = compute_rmsle(y_test.values, y_pred)
print("RMSLE on test set:", rmsle)
```
